# Greedy Algorithms

### What are Greedy Algorithms?
* any algorithm that makes a locally optimal decision at each step
    * e.g choosing the largest/smallest value in a given set

### Potential Signs That You Need to Try Greedy
* you're asked to get the highest/lowest sum or value from a simple array
    * beware that some cases may require more complicated algorithms like DP to get the absolute optimal answer
    

### Typical Implementations of Greedy Algorithms
* [heaps](../7-Trees/7.2.PriorityQueues&BinaryHeaps.ipynb) : typically O(log(N)) to O(N*log(N)) time complexity
* [monotonic stacks](../4-Linear_Data_Structures/4.1.DataStructures.ipynb): : potentially O(N) time complexity
* [sort algorithms](../6-Search&Sort/6.3.Sorting.ipynb): typically O(Nlog(N)) time complexity
